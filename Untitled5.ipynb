```python
# Mengubah kolom tanggal menjadi tipe datetime
df['tanggal'] = pd.to_datetime(df['tanggal'])

# Mapping harga ke nama menu
mapping_menu = {
    2000: "Kerupuk",
    3000: "Teh Hangat",
    4000: "Es Teh",
    5000: "Es Jeruk",
    8000: "Roti Bakar",
    10000: "Mie Ayam",
    12000: "Nasi Goreng"
}

# Mengisi menu yang kosong berdasarkan harga
df['menu'] = df['menu'].fillna(df['harga'].map(mapping_menu))

# Mengisi data terjual yang kosong dengan median berdasarkan menu
df['terjual'] = df['terjual'].fillna(
    df.groupby('menu')['terjual'].transform('median')
)

# Mengubah terjual menjadi bilangan bulat
df['terjual'] = df['terjual'].astype(int)

# Mengecek kembali missing value
df.isnull().sum()
```

```python
# Menghapus data yang duplikat
df = df.drop_duplicates()

# Mengecek jumlah data setelah duplikat dihapus
print("Jumlah data setelah cleaning:", len(df))

# Mengecek apakah masih ada duplikat
print("Jumlah duplikat:", df.duplicated().sum())
```

```python
# Menampilkan data yang jumlah terjualnya lebih dari 30
filter_data = df[df['terjual'] > 30]

filter_data
```

```python
# Mengurutkan data berdasarkan jumlah terjual
# dari terbesar ke terkecil

sorting_data = df.sort_values(
    by='terjual',
    ascending=False
)

sorting_data
```

```python
# Membuat kolom total_penjualan
# Rumus: harga x jumlah terjual

df['total_penjualan'] = df['harga'] * df['terjual']

# Menampilkan beberapa data
df.head()
```

```python
# Mengelompokkan data berdasarkan menu
group_menu = df.groupby('menu').agg(
    jumlah_terjual=('terjual', 'sum'),
    pendapatan=('total_penjualan', 'sum')
)

# Mengurutkan berdasarkan pendapatan terbesar
group_menu = group_menu.sort_values(
    by='pendapatan',
    ascending=False
)

group_menu
```
